```python

import pandas as pd


def analyze_movies(df):
    """Complete movie analysis function with fix for duplicate index reindexing."""
    results = {}

    # Ensure Gross_Millions exists or calculate default
    if "Gross_Millions" not in df.columns and "Gross" in df.columns:
        df["Gross_Millions"] = (
            df["Gross"]
            .astype(str)
            .str.replace("$", "", regex=False)
            .str.replace("M", "", regex=False)
            .astype(float)
        )

    # 1. Basic statistics
    results["total_movies"] = len(df)
    results["avg_rating"] = df["Rating"].mean()
    results["total_gross"] = (
        df["Gross_Millions"].sum() if "Gross_Millions" in df.columns else 0.0
    )

    # 2. Top performers
    top_movie_df = df.nlargest(1, "Rating")
    if not top_movie_df.empty:
        results["top_movie"] = top_movie_df[["Title", "Rating"]].to_dict(
            orient="records"
        )[0]
    else:
        results["top_movie"] = {"Title": "N/A", "Rating": None}

    results["top_director"] = (
        df.groupby("Director")["Rating"].mean().idxmax()
        if "Director" in df.columns
        else "N/A"
    )

    # 3. Yearly analysis
    agg_dict = {"Rating": ["mean", "count"]}
    if "Gross_Millions" in df.columns:
        agg_dict["Gross_Millions"] = "sum"

    results["yearly_stats"] = df.groupby("Year").agg(agg_dict).round(2)

    # 4. Genre analysis (FIX: Using df.explode directly to avoid duplicate index issues)
    if "Genre" in df.columns:
        # Create a copy with split genres
        df_exploded = df.copy()
        df_exploded["Genre"] = df_exploded["Genre"].str.split(",")
        df_exploded = df_exploded.explode("Genre")
        df_exploded["Genre"] = df_exploded["Genre"].str.strip()

        results["top_genre"] = df_exploded["Genre"].value_counts().idxmax()
        results["genre_ratings"] = (
            df_exploded.groupby("Genre")["Rating"]
            .mean()
            .sort_values(ascending=False)
        )
    else:
        results["top_genre"] = "N/A"
        results["genre_ratings"] = pd.Series(dtype=float)

    # 5. Correlations
    results["rating_votes_corr"] = (
        df["Rating"].corr(df["Votes"]) if "Votes" in df.columns else None
    )

    return results


# Run analysis
analysis_results = analyze_movies(df)

# Print results
print("=== MOVIE DATASET ANALYSIS ===\n")
print(f"Total Movies: {analysis_results['total_movies']}")
print(f"Average Rating: {analysis_results['avg_rating']:.2f}")
print(f"Total Gross: ${analysis_results['total_gross']:.0f}M")
print(f"Top Director: {analysis_results['top_director']}")
print(f"Most Common Genre: {analysis_results['top_genre']}")
if analysis_results["rating_votes_corr"] is not None:
    print(
        f"Rating - Votes Correlation: {analysis_results['rating_votes_corr']:.3f}"
    )

print("\nYearly Statistics:")
print(analysis_results["yearly_stats"])

print("\nTop 5 Genres by Rating:")
print(analysis_results["genre_ratings"].head())
```

```python
import pandas as pd


def clean_movie_data(df):
    """Comprehensive data cleaning pipeline."""
    df_clean = df.copy()

    # 1. Clean string columns
    df_clean["Title"] = df_clean["Title"].str.strip()
    df_clean["Director"] = df_clean["Director"].str.title()
    df_clean["Actor"] = df_clean["Actor"].str.strip()

    # 2. Clean numeric columns
    df_clean["Gross_Millions"] = (
        df_clean["Gross"].str.replace("$", "", regex=False).str.replace("M", "", regex=False).astype(float)
    )

    df_clean["Runtime_Minutes"] = (
        df_clean["Runtime"].str.replace(" min", "", regex=False).astype(int)
    )

    # 3. Create categorical columns
    rating_bins = [0, 7, 8, 9, 10]
    rating_labels = ["Poor", "Average", "Good", "Excellent"]
    df_clean["Rating_Category"] = pd.cut(
        df_clean["Rating"], bins=rating_bins, labels=rating_labels
    )

    # 4. Split genres for analysis
    df_clean["Genre_List"] = df_clean["Genre"].str.split(",")
    df_clean["Primary_Genre"] = df_clean["Genre_List"].str[0]

    # 5. Add derived columns
    df_clean["Rating_Rounded"] = df_clean["Rating"].round(1)
    df_clean["Votes_Millions"] = df_clean["Votes"] / 1_000_000

    # 6. Handle missing values
    df_clean["Rating_Category"] = df_clean["Rating_Category"].cat.add_categories("Not Rated")
    df_clean = df_clean.fillna({
        "Rating_Category": "Not Rated",
        "Gross_Millions": df_clean["Gross_Millions"].median(),
    })

    return df_clean


# Apply cleaning pipeline
df_cleaned = clean_movie_data(df)
print("Cleaned DataFrame info:")
print(df_cleaned.info())
print("\nFirst row after cleaning:")
print(df_cleaned.iloc[0])
```

```python
# Handle KeyError
try:
    _ = df["NonExistentColumn"]
except KeyError as e:
    print(f"Error: {e} - Column missing. Check column names with df.columns")

# Handle TypeError
try:
    _ = df["Gross"] + 100
except TypeError as e:
    print(f"Error: {e} - Cannot operate on strings. Convert with pd.to_numeric()")

# Handle ValueError with coercions
try:
    pd.to_numeric(df["Gross"], errors="raise")
except ValueError as e:
    print(f"Error: {e} - Non-numeric string found. Use errors='coerce'")

# Defensive checks
if "Rating" in df.columns:
    df["Rating"] = df["Rating"].fillna(df["Rating"].median())
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Generate dummy dataset
np.random.seed(42)
data = {
    "Title": [f"Movie_{i}" for i in range(1, 21)],
    "Rating": np.random.uniform(6, 9.5, 20).round(1),
    "Year": np.random.choice([2018, 2019, 2020, 2021], 20),
    "Votes": np.random.randint(50000, 250000, 20),
}

df = pd.DataFrame(data)

# Print summary
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nStatistics:\n", df.describe())

# Visualization Dashboard
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].hist(df["Rating"], bins=10, edgecolor="black")
axes[0, 0].set_title("Rating Distribution")

axes[0, 1].boxplot(df["Votes"])
axes[0, 1].set_title("Votes Boxplot")

axes[1, 0].scatter(df["Votes"], df["Rating"], alpha=0.6)
axes[1, 0].set_xlabel("Votes")
axes[1, 0].set_ylabel("Rating")
axes[1, 0].set_title("Rating vs Votes")

yearly_avg = df.groupby("Year")["Rating"].mean()
axes[1, 1].bar(yearly_avg.index, yearly_avg.values)
axes[1, 1].set_title("Average Rating by Year")

plt.tight_layout()
plt.show()
```
