```python
ds = []

# Correctly select the overall product containers first
mobiles_containers = soup.find_all('div', class_='jIjQ8S')

for mobile in mobiles_containers:
    try:
        name = mobile.find('div', class_='RG5Slk').text
        # Extracting product features
        features_list = mobile.find_all('li', class_='DTBslk')
        features = [f.text for f in features_list]

        ds.append({
            'Product_Name': name,
            'Specs': ', '.join(features)
        })
    except Exception as e:
        continue

df = pd.DataFrame(ds)
display(df.head())
```
